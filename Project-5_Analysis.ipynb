### 📱 Treatment of Missing IMEI Data

IMEI (International Mobile Equipment Identity) is a unique identifier used to identify a mobile device. Since it represents an identity rather than a measurable numerical value, it should not be handled like a continuous numeric variable.

**Why median imputation is not appropriate:**

* IMEI is an identifier, not a numerical measurement.
* A median value has no meaningful interpretation for an IMEI.
* Statistical imputation can generate an invalid or misleading device identifier.
* IMEI values should remain linked to the correct devices.

**Recommended Approach:**

Missing IMEI values should either be retained as missing (`NaN`) or handled using a suitable identifier-based method according to the business requirement and data source.

**Conclusion:**
Since IMEI is a unique device identifier rather than a measurable numeric feature, using median or mean imputation is not appropriate.

### 🛡️ Outlier Control & Data Capping

The **Start ms** and **End ms** variables were capped within the range of **0–999** to minimize the impact of extreme and invalid observations.

* **Lower limit:** 0
* **Upper limit:** 999
* Values below **0** were replaced with **0**.
* Values above **999** were replaced with **999**.

This approach helps maintain data consistency while reducing the influence of extreme values during analysis.


````python
# Select numerical columns for outlier treatment
outlier_cols = df.select_dtypes("number").columns

outlier_summary = []

for c in outlier_cols:
    q1, q3 = df[c].quantile([.25, .75])
    iqr = q3 - q1
    lo, hi = q1 - 1.5*iqr, q3 + 1.5*iqr
    n = ((df[c] < lo) | (df[c] > hi)).sum()
    df[c] = df[c].clip(lo, hi)
    outlier_summary.append([c, n, lo, hi])

outlier_summary_df = pd.DataFrame(
    outlier_summary,
    columns=["Column", "Outliers Before", "Lower Bound", "Upper Bound"]
)

outlier_summary_df
````

````python
# IQR outlier detection
num_cols = df.select_dtypes(include="number").columns
outlier_data = []

for c in num_cols:
    q1, q3 = df[c].quantile([.25, .75])
    iqr = q3 - q1
    n = ((df[c] < q1-1.5*iqr) | (df[c] > q3+1.5*iqr)).sum()
    outlier_data.append([c, n])

# Outlier table
outlier_table = pd.DataFrame(
    outlier_data, columns=["Feature", "Outlier Count"]
)

outlier_table
````

````python
dfv=df[df['MSISDN/Number'].notna()&(df['MSISDN/Number'].astype(str).str.lower()!='unknown')].copy()

c=['Dur. (ms)','Total DL (Bytes)','Total UL (Bytes)']
dfv[c]=dfv[c].apply(pd.to_numeric,errors='coerce')
u=dfv.dropna(subset=c).groupby('MSISDN/Number').agg(
D=('Dur. (ms)','sum'),DL=('Total DL (Bytes)','sum'),UL=('Total UL (Bytes)','sum')).reset_index()

u['Sec']=u.D/1000
u['GB']=(u.DL+u.UL)/1024**3
u['Decile']=pd.qcut(u.Sec.rank(method='first'),10,labels=False)

d=u.groupby('Decile').agg(Users=('MSISDN/Number','count'),
Min_Duration=('Sec','min'),Max_Duration=('Sec','max'),Data_GB=('GB','sum')).reset_index()
top=d[d.Decile.between(6,10)]

print("Total Users:",len(u))
display(d.round(2)); display(top.round(2))

sns.barplot(data=top,x='Decile',y='Data_GB',hue='Decile',
            palette=['#FF007F','#FF1493','#C71585','#9C27B0','#6A1B9A'],
            legend=False)
plt.title('Data Consumption by Duration Deciles',fontweight='bold')
plt.xlabel('Duration Decile'); plt.ylabel('Data Usage (GB)')
plt.tight_layout(); plt.show()
````

````python
# PCA Features
pca_features = [
    'Dur. (ms)', 'Avg RTT DL (ms)', 'Avg RTT UL (ms)',
    'Avg Bearer TP DL (kbps)', 'Avg Bearer TP UL (kbps)',
    'TCP DL Retrans. Vol (Bytes)', 'TCP UL Retrans. Vol (Bytes)',
    'Total DL (Bytes)', 'Total UL (Bytes)',
    'Social Media Total (Bytes)', 'Google Total (Bytes)',
    'Email Total (Bytes)', 'YouTube Total (Bytes)',
    'Netflix Total (Bytes)', 'Gaming Total (Bytes)',
    'Other Total (Bytes)'
]

# Prepare Data
data = df[pca_features].apply(pd.to_numeric, errors='coerce')
data = data.fillna(data.median())

# Standardize + PCA
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

scaled = StandardScaler().fit_transform(data)
pca = PCA().fit(scaled)

# Explained Variance
variance = pd.DataFrame({
    'Component': [f'PC{i+1}' for i in range(len(pca.explained_variance_ratio_))],
    'Explained Variance (%)': pca.explained_variance_ratio_ * 100,
    'Cumulative Variance (%)': pca.explained_variance_ratio_.cumsum() * 100
})

display(variance.round(2))
````

## 🧠 Dimensionality Reduction & Telecom Feature Insights

### 🔹 PCA Variance Explained — Key Findings

🔹 **PC1 – 25.50%**
PC1 explains the largest share of variance, capturing the strongest overall pattern across telecom usage and network-related features.

🔹 **PC2 & PC3 – 12.50% & 9.68%**
Together with PC1, the first three components explain approximately **47.67%** of the total variance, capturing a substantial portion of the major patterns in the dataset.

🔹 **First 10 Components – 89.67%**
The first 10 principal components retain approximately **89.67% of the total variance**. This suggests that the original 16-dimensional feature space can potentially be reduced to 10 dimensions while preserving most of the information.

🔹 **PC11–PC15 & PC16**
PC11–PC15 contribute relatively small amounts of additional variance, while **PC16 contributes 0%**, suggesting that some features contain overlapping or redundant information.

### 💡 Overall Interpretation

PCA demonstrates that most of the variation in the original **16 features** can be represented using fewer dimensions. Retaining the first **10 principal components** preserves approximately **89.67% of the total variance**, providing a more compact representation while maintaining most of the dataset's information.

### 📋 What These 16 Features Represent

| Category                   | Features                                                     | Purpose                                                                 |
| -------------------------- | ------------------------------------------------------------ | ----------------------------------------------------------------------- |
| 📱 **Session Behaviour**   | Dur. (ms)                                                    | Measures user session duration                                          |
| 🌐 **Network Performance** | Avg RTT DL/UL, Avg Bearer TP DL/UL                           | Measures latency and data-transfer performance                          |
| 🔄 **Network Quality**     | TCP DL/UL Retrans. Vol                                       | Indicates retransmission behaviour and potential network-quality issues |
| 📊 **Data Consumption**    | Total DL/UL                                                  | Measures total uploaded and downloaded data                             |
| 🎮 **Application Usage**   | YouTube, Netflix, Gaming, Social Media, Google, Email, Other | Measures data consumption across application categories                 |

### ✅ Why These Features Are Suitable for PCA

These features are numerical and represent important aspects of user and network behaviour:

* ⏱️ **Session duration**
* 📶 **Network latency and throughput**
* 🔄 **TCP retransmission behaviour**
* 📥📤 **Data consumption**
* 📱 **Application-level usage**

Therefore, PCA can identify underlying patterns and relationships among these variables and represent them using a smaller number of principal components.

### 🚫 Why Identifiers Are Not Included

Identifiers such as **MSISDN/Number, IMSI, IMEI, and Bearer Id** are excluded because they primarily identify users, devices, or sessions rather than represent meaningful behavioural or network measurements.

Including such identifiers could introduce noise without contributing meaningful information to the dimensionality-reduction analysis.

````python
# User Engagement Analysis
cols = ['Dur. (ms)', 'Total DL (Bytes)', 'Total UL (Bytes)']

d = df[df['MSISDN/Number'].notna() &
       (df['MSISDN/Number'].astype(str).str.lower() != 'unknown')].copy()

d[cols] = d[cols].apply(pd.to_numeric, errors='coerce')
d = d.dropna(subset=cols)

d['Traffic'] = d['Total DL (Bytes)'] + d['Total UL (Bytes)']

user_engagement = d.groupby('MSISDN/Number').agg(
    Sessions=('MSISDN/Number', 'count'),
    Duration_ms=('Dur. (ms)', 'sum'),
    DL_Bytes=('Total DL (Bytes)', 'sum'),
    UL_Bytes=('Total UL (Bytes)', 'sum'),
    Traffic_Bytes=('Traffic', 'sum')
).reset_index()

user_engagement['Duration_sec'] = user_engagement['Duration_ms'] / 1000
user_engagement['Traffic_GB'] = user_engagement['Traffic_Bytes'] / 1024**3

display(user_engagement.head())
````

````python
# Import Libraries
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Engagement Features
features = ['Sessions', 'Duration_ms', 'Traffic_Bytes']

# Prepare Data
data = user_engagement[['MSISDN/Number'] + features].replace(
    [np.inf, -np.inf], np.nan
).dropna()

# Normalize + K-Means
X = StandardScaler().fit_transform(data[features])

data['Engagement_Cluster'] = KMeans(
    n_clusters=3, random_state=42, n_init=10
).fit_predict(X)

# Display Results
display(data.sort_values('Engagement_Cluster').head(10))
````

## 🎯 Finding the Ideal Cluster Count

The Elbow Method was applied to identify a suitable cluster count for the K-Means model. The inertia decreases rapidly as the number of clusters increases, but the rate of improvement becomes much smaller after **K = 4**.

This indicates that **4 clusters** provide a reasonable balance between model simplicity and within-cluster variation.

Therefore, the K-Means model was configured with **K = 4** to segment customers into four distinct engagement groups.


````python
# Set optimal K
optimal_k = 4

# Apply K-Means
final_kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)

# Assign clusters
data['Final_Cluster'] = final_kmeans.fit_predict(normalized_data)
````

# 4 Satisfaction Analysis

````python
# Engagement features
engagement_data = df.groupby('MSISDN/Number').agg(
    Sessions=('Bearer Id','count'),
    Duration=('Dur. (ms)','sum'),
    Traffic=('Total DL (Bytes)','sum')
).reset_index()

# Total traffic = DL + UL
engagement_data['Traffic'] += df.groupby(
    'MSISDN/Number'
)['Total UL (Bytes)'].sum().values

# Scale
eng_features = ['Sessions','Duration','Traffic']
engagement_scaled = StandardScaler().fit_transform(
    engagement_data[eng_features]
)

# First clustering
kmeans_engagement = KMeans(n_clusters=3, random_state=42, n_init=10)
engagement_data['Cluster'] = kmeans_engagement.fit_predict(
    engagement_scaled
)

# Engagement score
centers = kmeans_engagement.cluster_centers_
less_engaged = centers.mean(axis=1).argmin()

engagement_data['Engagement_Score'] = euclidean_distances(
    engagement_scaled, [centers[less_engaged]]
).ravel()
````

````python
# Experience score
centers = kmeans.cluster_centers_
worst = (centers[:,0] + centers[:,1] - centers[:,2]).argmax()

experience_data['Experience_Score'] = euclidean_distances(
    experience_scaled, [centers[worst]]
).ravel()
````

````python
# Match customer IDs
engagement_data['MSISDN/Number'] = (
    engagement_data['MSISDN/Number'].astype(str).str.replace(r'\.0$', '', regex=True)
)
experience_data['MSISDN/Number'] = (
    experience_data['MSISDN/Number'].astype(str).str.replace(r'\.0$', '', regex=True)
)

# Satisfaction score
satisfaction = engagement_data[
    ['MSISDN/Number', 'Engagement_Score']
].merge(
    experience_data[['MSISDN/Number', 'Experience_Score']],
    on='MSISDN/Number'
)

satisfaction['Satisfaction_Score'] = satisfaction[
    ['Engagement_Score', 'Experience_Score']
].mean(axis=1)

# Top 10 customers
display(satisfaction.nlargest(10, 'Satisfaction_Score').round(2))
````

• Average of both scores = Satisfaction Score

````python
# Imports
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
````

````python
# Regression model
data = engagement_data[['MSISDN/Number','Sessions','Duration','Traffic','Engagement_Score']].merge(
    experience_data[['MSISDN/Number','TCP','RTT','Throughput','Experience_Score']], on='MSISDN/Number'
)

data['Satisfaction_Score'] = data[['Engagement_Score','Experience_Score']].mean(axis=1)

X = data[['Sessions','Duration','Traffic','TCP','RTT','Throughput']]
y = data['Satisfaction_Score']

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=.2,random_state=42)

model = RandomForestRegressor(n_estimators=100,random_state=42)
model.fit(X_train,y_train)

y_pred = model.predict(X_test)

print("MAE:", mean_absolute_error(y_test,y_pred))
print("RMSE:", mean_squared_error(y_test,y_pred)**0.5)
print("R²:", r2_score(y_test,y_pred))
````
